## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

#### 6. Identifying outlier problem
Pada tahap ini, dilakukan pemeriksaan untuk mengidentifikasi nilai ekstrem (outlier) yang memiliki perbedaan cukup jauh dibandingkan dengan sebagian besar data lainnya. Pemeriksaan menggunakan metode Interquartile Range (IQR) dengan fokus pada atribut jumlah penyewaan sepeda untuk mengetahui adanya nilai yang berpotensi memengaruhi hasil analisis.

```python
numerical_columns = [
    'temp',
    'atemp',
    'hum',
    'windspeed',
    'casual',
    'registered',
    'cnt'
]

outlier_results = []

for column in numerical_columns:

    # Menghitung Q1, Q3, dan IQR
    Q1 = df_day[column].quantile(0.25)
    Q3 = df_day[column].quantile(0.75)
    IQR = Q3 - Q1

    # Menentukan batas bawah dan batas atas
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Mengidentifikasi outlier
    outliers = df_day[
        (df_day[column] < lower_bound) |
        (df_day[column] > upper_bound)
    ]

    # Menyimpan hasil identifikasi
    outlier_results.append({
        'Column': column,
        'Q1': round(Q1, 3),
        'Q3': round(Q3, 3),
        'IQR': round(IQR, 3),
        'Lower Bound': round(lower_bound, 3),
        'Upper Bound': round(upper_bound, 3),
        'Total Outliers': len(outliers)
    })

# Menampilkan hasil dalam bentuk tabel
outlier_df = pd.DataFrame(outlier_results)

outlier_df
```

Interpretasi:

Berdasarkan pemeriksaan menggunakan metode Interquartile Range (IQR), ditemukan potensi outlier pada tiga atribut, yaitu hum sebanyak 2 data, windspeed sebanyak 13 data, dan casual sebanyak 44 data. Sementara itu, atribut temp, atemp, registered, dan cnt tidak menunjukkan adanya outlier berdasarkan batas IQR.

Jumlah outlier terbanyak ditemukan pada kolom casual, yang menunjukkan adanya beberapa hari dengan jumlah penyewaan oleh pengguna casual jauh lebih tinggi dibandingkan sebagian besar hari lainnya. Kondisi tersebut dapat berkaitan dengan peningkatan permintaan penyewaan pada periode tertentu. Oleh karena itu, nilai-nilai tersebut perlu ditinjau lebih lanjut dan tidak langsung dihapus karena masih berpotensi merepresentasikan kondisi operasional yang sebenarnya.

### 2. Bike Rental Demand Segmentation Based on Season and Weather
Analisis lanjutan ini dilakukan untuk mengelompokkan tingkat permintaan penyewaan sepeda berdasarkan jumlah penyewaan harian. Teknik Binning digunakan untuk membagi jumlah penyewaan ke dalam tiga kategori, yaitu Low Demand, Medium Demand, dan High Demand. Selanjutnya, teknik Manual Grouping digunakan untuk mengidentifikasi distribusi kategori permintaan berdasarkan kombinasi musim dan kondisi cuaca. Hasil analisis ini dapat membantu memahami pola permintaan serta menjadi pertimbangan dalam perencanaan ketersediaan sepeda.

#### Categorizing bike rental demand
Pertama, jumlah penyewaan sepeda harian (cnt) dikelompokkan menjadi tiga kategori berdasarkan batas nilai yang ditentukan. Pada analisis ini, jumlah penyewaan hingga 3.000 dikategorikan sebagai Low Demand, 3.001–6.000 sebagai Medium Demand, dan di atas 6.000 sebagai High Demand.

```python
df_day['demand_category'] = pd.cut(
    df_day['cnt'],
    bins=[0, 3000, 6000, float('inf')],
    labels=['Low Demand', 'Medium Demand', 'High Demand']
)

demand_analysis = (
    df_day['demand_category']
    .value_counts()
    .reindex(['Low Demand', 'Medium Demand', 'High Demand'])
    .reset_index()
)

demand_analysis.columns = ['Demand Category', 'Total Days']

demand_analysis
```

#### Grouping demand based on season and weather
Selanjutnya, dilakukan pengelompokan berdasarkan kombinasi musim, kondisi cuaca, dan kategori permintaan. Pengelompokan ini bertujuan untuk mengetahui kondisi yang paling sering menghasilkan permintaan penyewaan sepeda pada setiap kategori.

```python
season_weather_demand = (
    df_day.groupby(
        ['season_name', 'weather_name', 'demand_category'],
        observed=False
    )
    .size()
    .reset_index(name='Total Days')
)

season_weather_demand = season_weather_demand[
    season_weather_demand['Total Days'] > 0
].sort_values('Total Days', ascending=False)

season_weather_demand
```

#### Visualizing demand segmentation
Untuk mempermudah interpretasi, distribusi kategori permintaan berdasarkan kombinasi musim dan kondisi cuaca divisualisasikan menggunakan stacked bar chart.

```python
df_day['season_weather'] = (
    df_day['season_name'] + ' - ' + df_day['weather_name']
)

demand_pivot = pd.crosstab(
    df_day['season_weather'],
    df_day['demand_category']
)

demand_pivot = demand_pivot[
    ['Low Demand', 'Medium Demand', 'High Demand']
]

demand_pivot.plot(
    kind='bar',
    stacked=True,
    figsize=(13, 7)
)

plt.title('Bike Rental Demand Distribution by Season and Weather')
plt.xlabel('Season and Weather')
plt.ylabel('Number of Days')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Demand Category')
plt.tight_layout()
plt.show()
```
